# Estructuras de control selectivas

```python
def obtenerMayor(param1, param2):
    if param1 < param2:
        print('{} es mayor que {}'.format(param2, param1))
```

```python
obtenerMayor(5,7)
```

```python
obtenerMayor(7, 5)
```

```python
x = y = z = 3
if x == y == z:
    print(True)
```

```python
def obtenerMayorv2(param1, param2):
    if param1 < param2:
        return param2
    else:
        return param1
```

```python
print("El mayor es {}".format(obtenerMayorv2(4, 20
                                             )))
```

```python
print("El mayor es {}".format(obtenerMayorv2(11, 6)))
```

```python
def obtenerMayor_idiom(param1, param2):
    valor = param2 if (param1 < param2) else param1
    return valor
```

```python
print("El mayor es {}".format(obtenerMayor_idiom(11, 6)))
```

```python
def numeros(num):
    if num == 1:
        print("Tu numero es 1")
    elif num == 2:
        print(" el numero es el 2")
    elif num == 3:
        print(" el numero es el 3")
    elif num == 4:
        print(" el numero es el 4")
    else:
        print("no hay opcion")
```

```python
numeros(2)
```

```python
numeros(5)
```

```python
def numeros_idiom(num):
    if num in (1,2,3,4):
        print("Tu numero es {}".format(num))
    else:
        print("{} no es una opcion.". format(num))
```

```python
numeros_idiom(2)
```

```python
numeros_idiom(5)
```

```python
def obtenerMasGrande(a, b, c):
    if a>b:
        if a>c:
            return a
        else:
            return c
    else:
        if b>c:
            return b
        else:
            return c
```

```python
print("El numero mas grande es {}". format(obtenerMasGrande(7,13,1)))
```

# Estructuras de control repetitivas

```python
def cuenta(limite):
    i = limite
    while True:
        print(i)
        i =i-1
        if i==0:
            break
```

```python
cuenta(10)
```

```python
def factorial(n):
    i=2
    tmp = 1
    while i < n+1:
        tmp = tmp * i 
        i = i + 1
    return tmp
```

```python
print(factorial(4))
```

```python
print(factorial(6))
```

```python
def cuenta_idiom(limite):
    for i in range(limite, 0, -1):
        print(i)
    else:
        print("Cuenta finalizada")
```

```python
cuenta_idiom(5)
```

```python
def cuenta_idiomv2(limite):
    for i in range(limite, 0, -1):
        print(i)
        if i == 3:
            break
    else:
        print("Cuenta finalizada")
```
